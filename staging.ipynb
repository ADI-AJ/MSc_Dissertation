```python
import pandas as pd
```

```python
# Loading data from csv file

df = pd.read_csv('../Source.csv')
```

```python
df.shape[0]
```

```python
#top 5 tecords of DF
df.head(5)
```

```python
#Dropping records where id and title is redundant
df = df.drop_duplicates(subset=['id'], keep='first')
df = df.drop_duplicates(subset=['original_title'], keep='first')
```

```python
df.shape[0]
```

Creating & storing Fact table

```python
#Creating & storing movie table
movie_cols = ['movie_id', 'original_title', 'averageRating', 'status', 'release_date', 'revenue', 'runtime', 'adult', 'budget', 'overview', 'poster_path']
df.rename(columns={'id': 'movie_id'}, inplace=True)
df_movie = df[movie_cols]
df_movie.to_csv('../Dataset/Tables/movie.csv', index=False)
```

Creating & storing dimension table

```python
#Creating & storing genre table with unique genre values and assigning genre_id
df_genres = df['genres'].str.split(',').explode().to_frame(name='genre')
df_genres['genre'] = df_genres['genre'].str.strip().str.title()
df_genres = df_genres.drop_duplicates(subset='genre', keep='first').reset_index(drop=True)
df_genres['genre_id'] = df_genres.index + 1
df_genres.to_csv('../Dataset/Tables/genre.csv', index=False)
```

```python
#Creating & storing production_companies table with unique company names and assigning company_id
df_companies = df['production_companies'].str.split(',').explode().to_frame(name='production_company')
df_companies['production_company'] = df_companies['production_company'].str.strip().str.title()
df_companies = df_companies.drop_duplicates(subset = 'production_company', keep='first').reset_index(drop=True)
df_companies['company_id'] = df_companies.index + 1
df_companies.to_csv('../Dataset/Tables/company.csv', index=False)
```

```python
#Creating & storing languages table with unique language names and assigning language_id
df_languages = df['spoken_languages'].str.strip().str.split(',').explode().to_frame(name='language')
df_languages['language'] = df_languages['language'].str.strip().str.title()
df_languages = df_languages.drop_duplicates(subset = 'language', keep='first').reset_index(drop=True)
df_languages['language_id'] = df_languages.index + 1
df_languages.to_csv('../Dataset/Tables/language.csv', index=False)
```

```python
#Creating & storing director table with unique directors names and assigning director_id
df_directors = df['directors'].str.split(',').explode().to_frame(name='director')
df_directors['director'] = df_directors['director'].str.strip().str.title()
df_directors = df_directors.drop_duplicates(subset = 'director', keep='first').reset_index(drop=True)
df_directors['director_id'] = df_directors.index + 1
df_directors.to_csv('../Dataset/Tables/director.csv', index=False)
```

```python
#Creating & storing writer table with unique writers names and assigning writer_id
df_writers = df['writers'].str.split(',').explode().to_frame(name='writer')
df_writers['writer'] = df_writers['writer'].str.strip().str.title()
df_writers = df_writers.drop_duplicates(subset = 'writer', keep='first').reset_index(drop=True)
df_writers['writer_id'] = df_writers.index + 1
df_writers.to_csv('../Dataset/Tables/writer.csv', index=False)
```

```python
#Creating & storing cast table with unique casts name and assigning cast_id
df_casts = df['cast'].str.split(',').explode().to_frame(name='cast')
df_casts['cast'] = df_casts['cast'].str.strip().str.title()
df_casts = df_casts.drop_duplicates(subset = 'cast', keep='first').reset_index(drop=True)
df_casts['cast_id'] = df_casts.index + 1
df_casts.to_csv('../Dataset/Tables/cast.csv', index=False)
```

Creating & storing relationship tables

```python
#Creating & storing movie & genre relationship table
df_movie_genre = df[['movie_id','genres']].copy()
df_movie_genre['genres'] = df['genres'].str.split(',')
df_movie_genre = df_movie_genre.explode('genres').reset_index(drop=True)
df_movie_genre['genres'] = df_movie_genre['genres'].str.strip().str.title()
df_movie_genre = pd.merge(df_movie_genre, df_genres, how='left', left_on='genres', right_on='genre').drop(['genres','genre'],axis='columns').reset_index(drop=True)
df_movie_genre = df_movie_genre.drop_duplicates(subset=['movie_id','genre_id'], keep='first').reset_index(drop=True)
df_movie_genre.to_csv('../Dataset/Relationship_tables/movie_genre.csv', index=False)
```

```python
#Creating & storing movie & production company relationship table
df_movie_company = df[['movie_id','production_companies']].copy()
df_movie_company['production_companies'] = df['production_companies'].str.split(',')
df_movie_company = df_movie_company.explode('production_companies').reset_index(drop=True)
df_movie_company['production_companies'] = df_movie_company['production_companies'].str.strip().str.title()
df_movie_company = pd.merge(df_movie_company, df_companies, how='left', left_on='production_companies', right_on='production_company').drop(['production_companies','production_company'],axis='columns').reset_index(drop=True)
df_movie_company = df_movie_company.drop_duplicates(subset=['movie_id','company_id'], keep='first').reset_index(drop=True)
df_movie_company.to_csv('../Dataset/Relationship_tables/movie_company.csv', index=False)
```

```python
#Creating & storing movie & language relationship table
df_movie_language = df[['movie_id','spoken_languages']].copy()
df_movie_language['spoken_languages'] = df['spoken_languages'].str.split(',')
df_movie_language = df_movie_language.explode('spoken_languages').reset_index(drop=True)
df_movie_language['spoken_languages'] = df_movie_language['spoken_languages'].str.strip().str.title()
df_movie_language = pd.merge(df_movie_language, df_languages, how='left', left_on='spoken_languages', right_on='language').drop(['spoken_languages','language'],axis='columns').reset_index(drop=True)
df_movie_language = df_movie_language.drop_duplicates(subset=['movie_id','language_id'], keep='first').reset_index(drop=True)
df_movie_language.to_csv('../Dataset/Relationship_tables/movie_language.csv', index=False)
```

```python
#Creating & storing movie & directors relationship table
df_movie_director = df[['movie_id','directors']].copy()
df_movie_director['directors'] = df['directors'].str.split(',')
df_movie_director = df_movie_director.explode('directors').reset_index(drop=True)
df_movie_director['directors'] = df_movie_director['directors'].str.strip().str.title()
df_movie_director = pd.merge(df_movie_director, df_directors, how='left', left_on='directors', right_on='director').drop(['directors','director'],axis='columns').reset_index(drop=True)
df_movie_director = df_movie_director.drop_duplicates(subset=['movie_id','director_id'], keep='first').reset_index(drop=True)
df_movie_director.to_csv('../Dataset/Relationship_tables/movie_director.csv', index=False)
```

```python
#Creating & storing movie & writers relationship table
df_movie_writer = df[['movie_id','writers']].copy()
df_movie_writer['writers'] = df['writers'].str.split(',')
df_movie_writer = df_movie_writer.explode('writers').reset_index(drop=True)
df_movie_writer['writers'] = df_movie_writer['writers'].str.strip().str.title()
df_movie_writer = pd.merge(df_movie_writer, df_writers, how='left', left_on='writers', right_on='writer').drop(['writers','writer'],axis='columns').reset_index(drop=True)
df_movie_writer = df_movie_writer.drop_duplicates(subset=['movie_id','writer_id'], keep='first').reset_index(drop=True)
df_movie_writer.to_csv('../Dataset/Relationship_tables/movie_writer.csv', index=False)
```

```python
#Creating & storing movie & cast relationship table
df_movie_cast = df[['movie_id','cast']].copy()
df_movie_cast['cast'] = df['cast'].str.split(',')
df_movie_cast = df_movie_cast.explode('cast').reset_index(drop=True)
df_movie_cast['cast'] = df_movie_cast['cast'].str.strip().str.title()
df_movie_cast = pd.merge(df_movie_cast, df_casts, how='left', left_on='cast', right_on='cast').drop(['cast','cast'],axis='columns').reset_index(drop=True)
df_movie_cast = df_movie_cast.drop_duplicates(subset=['movie_id','cast_id'], keep='first').reset_index(drop=True)
df_movie_cast.to_csv('../Dataset/Relationship_tables/movie_cast.csv', index=False)
```
