```python
import pandas as pd
```

```python
# Loading data from csv file

df = pd.read_csv('Source.csv')
```

```python
df.shape[0]
```

```python
#top 5 tecords of DF
df.head(5)
```

```python
#Dropping records where id and title is redundant
df = df.drop_duplicates(subset=['id'], keep='first')
df = df.drop_duplicates(subset=['original_title'], keep='first')
```

```python
df.shape[0]
```

Creating & storing Fact table

```python
#Creating & storing movie table
movie_cols = ['movie_id', 'original_title', 'averageRating', 'status', 'release_date', 'revenue', 'runtime', 'adult', 'budget', 'overview', 'poster_path']
df.rename(columns={'id': 'movie_id'}, inplace=True)
df_movie = df[movie_cols]
df_movie.to_csv('Dataset/Tables/movie.csv', index=False)
```

Creating & storing dimension table

```python
#Creating & storing genre table with unique genre values and assigning genre_id
df_genres = df['genres'].str.split(',').explode().to_frame(name='genre')
df_genres['genre'] = df_genres['genre'].str.strip()
df_genres['_genres'] = df_genres['genre'].str.title()
df_genres = df_genres.drop_duplicates(subset = '_genres', keep='first').reset_index(drop=True)
df_genres = df_genres.drop(columns = '_genres')
df_genres['genre_id'] = df_genres.index + 1
df_genres.to_csv('Dataset/Tables/genre.csv', index=False)
```

```python
#Creating & storing production_companies table with unique company names and assigning company_id
df_companies = df['production_companies'].str.split(',').explode().to_frame(name='production_company')
df_companies['production_company'] = df_companies['production_company'].str.strip()
df_companies['_production_companies'] = df_companies['production_company'].str.title()
df_companies = df_companies.drop_duplicates(subset = '_production_companies', keep='first').reset_index(drop=True)
df_companies = df_companies.drop(columns = '_production_companies')
df_companies['company_id'] = df_companies.index + 1
df_companies.to_csv('Dataset/Tables/company.csv', index=False)
```

```python
#Creating & storing languages table with unique language names and assigning language_id
df_languages = df['spoken_languages'].str.strip().str.split(',').explode().to_frame(name='language')
df_languages['language'] = df_languages['language'].str.strip()
df_languages['_languages'] = df_languages['language'].str.title()
df_languages = df_languages.drop_duplicates(subset = '_languages', keep='first').reset_index(drop=True)
df_languages = df_languages.drop(columns = '_languages')
df_languages['language_id'] = df_languages.index + 1
df_languages.to_csv('Dataset/Tables/language.csv', index=False)
```

```python
#Creating & storing director table with unique directors names and assigning director_id
df_directors = df['directors'].str.split(',').explode().to_frame(name='director')
df_directors['director'] = df_directors['director'].str.strip()
df_directors['_directors'] = df_directors['director'].str.strip().str.title()
df_directors = df_directors.drop_duplicates(subset = '_directors', keep='first').reset_index(drop=True)
df_directors = df_directors.drop(columns = '_directors')
df_directors['director_id'] = df_directors.index + 1
df_directors.to_csv('Dataset/Tables/director.csv', index=False)
```

```python
#Creating & storing writer table with unique writers names and assigning writer_id
df_writers = df['writers'].str.split(',').explode().to_frame(name='writer')
df_writers['writer'] = df_writers['writer'].str.strip()
df_writers['_writers'] = df_writers['writer'].str.title()
df_writers = df_writers.drop_duplicates(subset = '_writers', keep='first').reset_index(drop=True)
df_writers = df_writers.drop(columns = '_writers')
df_writers['writer_id'] = df_writers.index + 1
df_writers.to_csv('Dataset/Tables/writer.csv', index=False)
```

```python
#Creating & storing cast table with unique casts name and assigning cast_id
df_casts = df['cast'].str.split(',').explode().to_frame(name='cast')
df_casts['cast'] = df_casts['cast'].str.strip()
df_casts['_casts'] = df_casts['cast'].str.title()
df_casts = df_casts.drop_duplicates(subset = '_casts', keep='first').reset_index(drop=True)
df_casts = df_casts.drop(columns = '_casts')
df_casts['cast_id'] = df_casts.index + 1
df_casts.to_csv('Dataset/Tables/cast.csv', index=False)
```

Creating & storing relationship tables

```python
#Creating & storing movie & genre relationship table
df_movie_genre = df[['movie_id','genres']].copy()
df_movie_genre['genres'] = df['genres'].str.split(',')
df_movie_genre = df_movie_genre.explode('genres').reset_index(drop=True)
df_movie_genre['genres'] = df_movie_genre['genres'].str.strip()
df_movie_genre = pd.merge(df_movie_genre, df_genres, how='left', left_on='genres', right_on='genre').drop(['genres','genre'],axis='columns').reset_index(drop=True)
df_movie_genre.to_csv('Dataset/Relationship_tables/movie_genre.csv', index=False)
```

```python
#Creating & storing movie & production company relationship table
df_movie_company = df[['movie_id','production_companies']].copy()
df_movie_company['production_companies'] = df['production_companies'].str.split(',')
df_movie_company = df_movie_company.explode('production_companies').reset_index(drop=True)
df_movie_company['production_companies'] = df_movie_company['production_companies'].str.strip()
df_movie_company = pd.merge(df_movie_company, df_companies, how='left', left_on='production_companies', right_on='production_company').drop(['production_companies','production_company'],axis='columns').reset_index(drop=True)
df_movie_company.to_csv('Dataset/Relationship_tables/movie_company.csv', index=False)
```
